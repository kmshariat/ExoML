```python
import pandas as pd
from functools import reduce

files = {
    "AdaBoost": "predictions_tess_adaboost_th0.5.csv",
    "DecisionTree": "predictions_tess_decisiontree_th0.5.csv",
    "kNN": "predictions_tess_kNN_th0.5.csv",
    "RF": "predictions_tess_RF_th0.5.csv",
    "XGBoost": "predictions_tess_XGBoost_th0.5.csv",
}

dfs = []

for model, file in files.items():
    df = pd.read_csv(file)

    # False positives only
    df = df[(df["ttv_flag"] == 0) & (df["ttv_predicted"] == 1)]

    dfs.append(
        df[["pl_name"]].assign(**{model: 1})
    )

# Merge all planets appearing in any model's false positives
merged = reduce(
    lambda left, right: pd.merge(left, right, on="pl_name", how="outer"),
    dfs
).fillna(0)

# Number of models that falsely predicted TTV
model_cols = list(files.keys())
merged["n_models"] = merged[model_cols].sum(axis=1)

print(f"All 5 models : {(merged['n_models'] == 5).sum()}")
print(f"Any 4 models : {(merged['n_models'] >= 4).sum()}")
print(f"Any 3 models : {(merged['n_models'] >= 3).sum()}")
print(f"Any 2 models : {(merged['n_models'] >= 2).sum()}")
print(f"Any 1 model  : {(merged['n_models'] >= 1).sum()}")
```

```python
for n in [5, 4, 3, 2, 1]:
    planets = merged.loc[merged["n_models"] == n, "pl_name"].tolist()
    print(f"\nExactly {n} models ({len(planets)} planets)")
    print(planets)
```
